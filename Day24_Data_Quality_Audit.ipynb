```python
import pandas as pd

df = pd.read_excel("GlobalSuperstore.xlsx", sheet_name="Orders")

print("Dataset loaded successfully!")
```

```python
# Display first 5 rows
df.head()
```

```python
# Check dataset dimensions
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

# Display column names
print(df.columns.tolist())
```

```python
# Check missing values in each column

missing_values = df.isnull().sum()

print("Missing Values in Each Column:")
print(missing_values)
```

```python
# Calculate missing value percentage

missing_percentage = (df.isnull().sum() / len(df)) * 100

print(missing_percentage.round(2))
```

```python
# Count duplicate records

duplicate_count = df.duplicated().sum()

print("Total Duplicate Rows:", duplicate_count)
```

```python
# Display duplicate rows

duplicates = df[df.duplicated()]

duplicates.head()
```

```python
duplicate_percentage = (duplicate_count / len(df)) * 100

print("Duplicate Percentage:", round(duplicate_percentage, 2), "%")
```

```python
# Display statistical summary of numerical columns

df.describe()
```

```python
invalid_sales = df[df['Sales'] < 0]

print("Invalid Sales Records:", len(invalid_sales))
```

```python
invalid_quantity = df[df['Quantity'] <= 0]

print("Invalid Quantity Records:", len(invalid_quantity))
```

```python
invalid_discount = df[
    (df['Discount'] < 0) | (df['Discount'] > 1)
]

print("Invalid Discount Records:", len(invalid_discount))
```

```python
text_columns = df.select_dtypes(include='object').columns

for col in text_columns:
    extra_spaces = df[col].astype(str).str.strip().ne(df[col].astype(str)).sum()
    if extra_spaces > 0:
        print(col, ":", extra_spaces, "values with extra spaces")
```

```python
df_clean = df.copy()

print("Original Dataset:", df.shape)
print("Working Dataset:", df_clean.shape)
```

```python
df_clean = df_clean.drop_duplicates()

print("Dataset after removing duplicates:", df_clean.shape)
```

```python
df_clean.isnull().sum()
```

```python
# Fill missing Postal Codes with 0
if 'Postal Code' in df_clean.columns:
    df_clean['Postal Code'] = df_clean['Postal Code'].fillna(0)

print("Missing values handled in Postal Code.")
```

```python
text_columns = df_clean.select_dtypes(include='object').columns

for col in text_columns:
    df_clean[col] = df_clean[col].str.strip()

print("Extra spaces cleaned successfully!")
```

```python
print("Remaining duplicates:", df_clean.duplicated().sum())

print("\nRemaining missing values:")
print(df_clean.isnull().sum())
```

```python
comparison = pd.DataFrame({
    'Metric': [
        'Total Rows',
        'Total Columns',
        'Duplicate Records',
        'Missing Values'
    ],
    'Before Cleaning': [
        df.shape[0],
        df.shape[1],
        df.duplicated().sum(),
        df.isnull().sum().sum()
    ],
    'After Cleaning': [
        df_clean.shape[0],
        df_clean.shape[1],
        df_clean.duplicated().sum(),
        df_clean.isnull().sum().sum()
    ]
})

comparison
```

```python
audit_report = pd.DataFrame({
    'Column': df.columns,
    'Missing Values': df.isnull().sum().values,
    'Missing Percentage': (
        df.isnull().mean().values * 100
    ).round(2),
    'Data Type': df.dtypes.astype(str).values,
    'Unique Values': df.nunique().values
})

audit_report
```

```python
with pd.ExcelWriter('Day24_Data_Quality_Audit_Report.xlsx') as writer:

    df_clean.to_excel(
        writer,
        sheet_name='Cleaned Data',
        index=False
    )

    comparison.to_excel(
        writer,
        sheet_name='Comparison',
        index=False
    )

    audit_report.to_excel(
        writer,
        sheet_name='Audit Report',
        index=False
    )

print("Excel Report Generated Successfully!")
```
